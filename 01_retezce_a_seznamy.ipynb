# 1. Proměnné v paměti: Mutabilita a reference

-**Reference v paměti**: proměnná je odkaz na objek v paměti (`id(x)`)
-**Imutabilní (neměnné)**: `int`, `float`, `bool`, `str`, `tuple` (úprava vytvoří nový objekt)
-**Mutabilní (měnitelné)**: `list`, `dict`, `set`, (změna probíhá přímo na místě / *in-place*)
-**Pozor na alias**: `b = a` nevytváří kopii, obě proměnné ukazují na stejný objekt, skutečná kopie: `a.copy()` nebo `a[:]`

```python
# x1 a x2 jsou dva různé objekty
x = 10
print(f"Původní x={x}, id={id(x)}")
x += 1
print(f"Původní x={x}, id={id(x)}")

a = [1, 2, 3]
b = a
b.append(4)
print(f"a: {a}, b: {b}") # Do a by se neměla přidat ta 4, ale je tam

c = a.copy()
c.append(99)
print(f"a: {a}, c (kopie): {c}")
```

## 2. Řetězce (`str`) jako neměnná sekvence

-**Indexování**: od `0` (zepředu), `-1` (odzadu)
-**Řezy**: `sekvence[od:do:krok]` (mez `do` je vyjma, `krok` je volitelný, výchozí = 1)
-**Obrácení**: `text[::-1]`
-**Základní metody**: `.strip()`, `.lower()`, `.upper()`, `.replace()`
-**Převod text <-> seznam**: `.split(oddelovac)` (rozdělí), `oddelovac.join(seznam)` (spojení)

```python
text = "  Gymnázium Nad Štolou"
cisty = text.strip()
print(f"Ořezaný: '{cisty}'")

slovo = "Informatika"
print(slovo[0:4])
print(slovo[-4:])
print(slovo[::2])
print(slovo[::-1])

veta = "jablko, hruska, banan, pomeranc"
ovoce = veta.split(", ")
print(f"Seznam: {ovoce}")
znovu_spojene = " | ".join(ovoce)
print(f"Spojeno: {znovu_spojene}")
```

## Seznamy (`list`) a generátorová notace

- **Metody**: `.append()` (přidání na konec), `.insert(i, x)` (vložení), `.pop(i)` (vyjmutí prvku), `.remove(x)` (odstranění hodnoty)
- **Řazení**: `seznam.sort()` (in-place změna) vs. `sorted(seznam)` (vrací nový seznam)
- **List comprehension** (generátorová notace seznamu): `[výraz for prvek in kolekce if podmínka]`
- **Pozor na 2D seznamy**: `[[0] * n] * n` vytvoří n odkazů na tentýž řádek, lepší alternativa `[[0 for _ in range(n)] for _ in range(n)]`

```python
cisla = [7, 2, 9, 1, 5, 2]

cisla.append(10)
vyjmute = cisla.pop(0) # vyndám prvek indexu 0
print(f"Vyjmuto: {vyjmute}, zbytek: {cisla}")

serazeno = sorted(cisla)
print(f"Seřazené: {serazeno}, původní: {cisla}")
cisla.sort(reverse=True)
print(f"Úvodní seřazené opačně: {cisla}")

druhe_mocniny_lichych = [x**2 for x in serazeno if x % 2 == 1]
print(f"Mocniny lichých čísel: {druhe_mocniny_lichych}")

mrizka = [[0 for _ in range(3)] for _ in range(3)]
mrizka[1][1] = 5
print(f"Mřížka 3x3: {mrizka}")
```

Mikroůloha 2: Pomocí list comprehension převeď teploty v °F na °C $(C = (F - 32) \cdot 5 / 9)$ zaokrouhlené na 1 desetinné místo (funkce `round`), pouze kladné teploty.

```python
teploty_f = [32, 68, 14, 85, 50, 23, 104]
 
teploty_c = [round((teploty_f[f] - 32) * 5 / 9) for f in range(0, len(teploty_f)) if teploty_f[f] >= 32]
print(teploty_c)
```
